## Instalujemy potrzebne biblioteki oraz inicjalizujemy tablice ocen studentów

```python
import numpy as np
import matplotlib.pyplot as plt
from statistics import mode
```

```python
student_scores = [78, 85, 92, 56, 69, 73, 88, 91, 45, 82, 67, 72, 93,
54, 77, 88, 60, 95, 81, 79, 90, 67, 65, 77, 71, 81, 85, 88, 79, 66, 72,
89, 64, 82, 79, 94, 70, 84, 66, 73]
```

# Zadanie 1: Histogram ocen studentów

```python
plt.hist(student_scores, bins=10, edgecolor='black')
plt.title('Histogram ocen studentów')
plt.xlabel('Ocena')
plt.ylabel('Częstotliwość')
plt.show()
```

# Zadanie 2: Identyfikacja wartosci odstajacych

```python
Q1 = np.percentile(student_scores, 25)
Q3 = np.percentile(student_scores, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = [
    score for score in student_scores
    if score < lower_bound or score > upper_bound
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Dolna granica:", lower_bound)
print("Gorna granica:", upper_bound)
print("Wartosci odstajace:", outliers)
```

# Zadanie 3: Miary statystyczne

```python
print("Srednia:", np.mean(student_scores))
print("Mediana:", np.median(student_scores))
print("Dominanta:", mode(student_scores))
print("Zakres:", np.ptp(student_scores))
print("Wariancja:", np.var(student_scores))
print("Odchylenie standardowe:", np.std(student_scores))
```
