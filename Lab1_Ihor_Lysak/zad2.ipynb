## Instalujemy potrzebne biblioteki oraz inicjalizujemy tablice ratingów

```python
import numpy as np
import matplotlib.pyplot as plt
from statistics import mode
```

```python
ratings = [
    3, 4, 5, 2, 3, 5, 4, 2, 3, 4,
    5, 1, 3, 5, 4, 2, 4, 3, 3, 5,
    4, 3, 2, 4, 5
]
```

# Zadanie 1: Wykres kołowy rozkładu ocen

```python
values, counts = np.unique(ratings, return_counts=True)

plt.pie(
    counts,
    labels=values,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Rozkład ocen pracowników')
plt.show()
```

# Zadanie 2: Miary tendencji centralnej i rozproszenia

```python
print("Srednia:", np.mean(ratings))
print("Mediana:", np.median(ratings))
print("Dominanta:", mode(ratings))
print("Zakres:", np.ptp(ratings))
print("Wariancja:", np.var(ratings))
print("Odchylenie standardowe:", np.std(ratings))
```
