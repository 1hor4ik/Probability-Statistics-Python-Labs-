## Instalujemy potrzebne biblioteki oraz inicjalizujemy tablice ratingów zadowolenia

```python
import numpy as np
import matplotlib.pyplot as plt
from statistics import mode
```

```python
satisfaction_ratings = [
    3, 4, 5, 2, 4, 3, 5, 1, 4, 2,
    3, 4, 5, 2, 3, 4, 5, 3, 2, 4
]
```

# Zadanie 1: Histogram ocen zadowolenia klientów

```python
plt.hist(
    satisfaction_ratings,
    bins=np.arange(0.5, 6.5, 1),
    edgecolor='black'
)

plt.title('Histogram ocen zadowolenia klientów')
plt.xlabel('Ocena zadowolenia')
plt.ylabel('Liczba klientów')
plt.xticks([1, 2, 3, 4, 5])
plt.show()
```

# Zadanie 2: Miary tendencji centralnej

```python
print("Srednia:", np.mean(satisfaction_ratings))
print("Mediana:", np.median(satisfaction_ratings))
print("Dominanta:", mode(satisfaction_ratings))
```

# Zadanie 3: Miary rozproszenia

```python
print("Wariancja:", np.var(satisfaction_ratings))
print("Odchylenie standardowe:", np.std(satisfaction_ratings))
```
