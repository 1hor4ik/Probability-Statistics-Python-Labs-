```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

# Dla rozkładu wykładniczego średnia wynosi 1 / λ. Gdy λ = 2, średnia teoretyczna wynosi 0.5, nie 1. W NumPy parametr scale oznacza średnią, więc ustawiamy scale=1/2.

```python
lambda_rate = 2
expected_mean = 1 / lambda_rate
values = np.random.exponential(scale=expected_mean, size=1000)
running_mean = np.cumsum(values) / np.arange(1, len(values) + 1)

plt.plot(running_mean, label="Średnia skumulowana")
plt.axhline(expected_mean, linestyle="--", label="Średnia teoretyczna = 0.5")
plt.xlabel("Liczba obserwacji")
plt.ylabel("Średnia")
plt.title("LLN – rozkład wykładniczy, lambda = 2")
plt.legend()
plt.grid(True)
plt.show()
```
