```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

# Losujemy 1000 wag z zakresu 0–10 gramów i liczymy średnią skumulowaną. Średnia teoretyczna rozkładu jednostajnego to (0 + 10) / 2 = 5 gramów.

```python
weights = np.random.uniform(0, 10, size=1000)
running_mean = np.cumsum(weights) / np.arange(1, len(weights) + 1)

plt.plot(running_mean, label="Średnia skumulowana")
plt.axhline(5, linestyle="--", label="Średnia teoretyczna = 5 g")
plt.xlabel("Liczba produktów")
plt.ylabel("Średnia waga (g)")
plt.title("LLN – waga produktów")
plt.legend()
plt.grid(True)
plt.show()
```
