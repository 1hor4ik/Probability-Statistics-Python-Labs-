```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

# Dla λ = 1 średnia rozkładu wykładniczego wynosi 1 / λ = 1 minutę. W NumPy ustawiamy więc scale=1.

```python
lambda_rate = 1
expected_mean = 1 / lambda_rate
waiting_times = np.random.exponential(scale=expected_mean, size=1000)
running_mean = np.cumsum(waiting_times) / np.arange(1, len(waiting_times) + 1)

plt.plot(running_mean, label="Średnia skumulowana")
plt.axhline(expected_mean, linestyle="--", label="Średnia teoretyczna = 1 minuta")
plt.xlabel("Liczba obserwacji")
plt.ylabel("Średni czas (minuty)")
plt.title("LLN – czas między połączeniami")
plt.legend()
plt.grid(True)
plt.show()
```
