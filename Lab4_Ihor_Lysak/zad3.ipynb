```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

# Rozkład Beta daje wartości od 0 do 1. Przekształcamy je do zakresu 10–60 minut: 10 + wynik * 50. Potem losujemy 40 prób po 25 wartości i liczymy ich średnie.

```python
beta_values = np.random.beta(2, 5, size=300)
training_times = 10 + beta_values * 50

sample_means = [
    np.mean(np.random.choice(training_times, size=25))
    for _ in range(40)
]

plt.hist(sample_means, bins=10, edgecolor="black")
plt.xlabel("Średni czas szkolenia (minuty)")
plt.ylabel("Częstość")
plt.title("Rozkład średnich z prób – czas szkolenia")
plt.show()
print("Średnia z 40 średnich:", round(np.mean(sample_means), 2))
```
