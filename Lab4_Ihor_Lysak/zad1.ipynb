# Prygotowujemy potrzebne biblioteki dla kolejnych zadań

```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

# Generujemy 10 000 wartości z rozkładu wykładniczego o średniej 500, a następnie losujemy z nich próbę 400 wartości.

```python
spending_population = np.random.exponential(scale=500, size=10000)
spending_sample = np.random.choice(spending_population, size=400, replace=False)

print("Średnia populacji:", round(np.mean(spending_population), 2))
print("Średnia próby:", round(np.mean(spending_sample), 2))
plt.hist(spending_sample, bins=20, edgecolor="black")
plt.xlabel("Miesięczne wydatki")
plt.ylabel("Częstość")
plt.title("Próba wydatków gospodarstw domowych")
plt.show()
```
