```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

# Generujemy 1000 wartości z rozkładu normalnego o średniej 2 g i wariancji 1 g². Odchylenie standardowe to pierwiastek z wariancji, czyli 1 g.

```python
weights_normal = np.random.normal(loc=2, scale=1, size=1000)
running_mean = np.cumsum(weights_normal) / np.arange(1, len(weights_normal) + 1)

plt.plot(running_mean, label="Średnia skumulowana")
plt.axhline(2, linestyle="--", label="Średnia teoretyczna = 2 g")
plt.xlabel("Liczba produktów")
plt.ylabel("Średnia waga (g)")
plt.title("LLN – rozkład normalny")
plt.legend()
plt.grid(True)
plt.show()
```
