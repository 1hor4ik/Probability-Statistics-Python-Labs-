# Importujemy wszystkie potrzebne biblioteki i utworzymy zmienne dotyczące wzrostu uczniów takich jak średni wzrost, odchylenie standartowe wzrostu i liczbę próbek do wygenerowania

```python
import numpy as np
import matplotlib.pyplot as plt

mean_height = 170
std_dev_height = 10
num_students = 1000
```

# Generujemy 1000 próbek wzrostu za pomocą komendy np.random.normal()

```python
heights = np.random.normal(
    loc=mean_height,
    scale=std_dev_height,
    size=num_students
)
```

# Obliczamy wartośc średnią i wariancję i rysujemy histogram

```python
sample_mean = np.mean(heights)
sample_variance = np.var(heights)

print("Średnia wysokość:", sample_mean)
print("Wariancja:", sample_variance)

plt.hist(heights, bins=30, edgecolor="black")
plt.xlabel("Wzrost (cm)")
plt.ylabel("Liczba uczniów")
plt.title("Rozkład wzrostu uczniów")
plt.show()
```
