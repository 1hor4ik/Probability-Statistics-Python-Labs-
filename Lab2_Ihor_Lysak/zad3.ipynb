# Importujemy wszystkie potrzebne biblioteki i utworzymy wszystkie potrzebne nam zmienne

```python

import numpy as np
import matplotlib.pyplot as plt

lambda_value = 3
num_samples = 500
```


# Generujemy 500 próbek liczby połączeń za pomocą funkcji np.random.poisson()


```python

calls = np.random.poisson(
    lam=lambda_value,
    size=num_samples
)
```


# Obliczamy wartość średnią i wariancję oraz rysujemy histogram

```python

sample_mean = np.mean(calls)
sample_variance = np.var(calls)

print("Średnia liczba połączeń:", sample_mean)
print("Wariancja:", sample_variance)

plt.hist(
    calls,
    bins=np.arange(-0.5, max(calls) + 1.5, 1),
    edgecolor="black"
)
plt.xlabel("Liczba połączeń w ciągu 10 minut")
plt.ylabel("Liczba obserwacji")
plt.title("Rozkład Poissona — połączenia w call center")
plt.grid()
plt.show()
```
