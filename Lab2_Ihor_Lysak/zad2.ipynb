# Importujemy wszystkie potrzebne biblioteki i utworzymy wszystkie potrzebne nam zmienne

```python

import numpy as np
import matplotlib.pyplot as plt

mean_temp = 5
std_dev_temp = 3
num_days = 300
```


# Generujemy 300 próbek temperatur za pomocą funkcji np.random.normal()

```python

temperatures = np.random.normal(
    loc=mean_temp,
    scale=std_dev_temp,
    size=num_days
)
```


# Obliczamy wartość średnią i wariancję oraz rysujemy histogram

```python

sample_mean = np.mean(temperatures)
sample_variance = np.var(temperatures)

print("Średnia temperatura:", sample_mean)
print("Wariancja:", sample_variance)

plt.hist(temperatures, bins=20, edgecolor="black")
plt.xlabel("Temperatura (°C)")
plt.ylabel("Liczba obserwacji")
plt.title("Rozkład temperatur zimą")
plt.grid()
plt.show()
```
