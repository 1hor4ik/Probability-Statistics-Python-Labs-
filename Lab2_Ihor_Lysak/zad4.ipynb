# Importujemy wszystkie potrzebne biblioteki i utworzymy wszystkie potrzebne nam zmienne

```python

import numpy as np
import matplotlib.pyplot as plt

lambda_value = 2
num_samples = 2000
```


# Generujemy 2000 próbek liczby literówek za pomocą funkcji np.random.poisson()

```python

typos = np.random.poisson(
    lam=lambda_value,
    size=num_samples
)
```


# Obliczamy wartość średnią i wariancję oraz rysujemy histogram

```python

sample_mean = np.mean(typos)
sample_variance = np.var(typos)

print("Średnia liczba literówek:", sample_mean)
print("Wariancja:", sample_variance)

plt.hist(
    typos,
    bins=np.arange(-0.5, max(typos) + 1.5, 1),
    edgecolor="black"
)
plt.xlabel("Liczba literówek na 50 stron")
plt.ylabel("Liczba obserwacji")
plt.title("Rozkład Poissona — liczba literówek")
plt.grid()
plt.show()
```
