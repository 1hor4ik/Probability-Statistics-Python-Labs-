```python
import pandas as pd
```

# Tworzymy ramkę danych na podstawie tabeli z instrukcji

```python
data = {
    "Yoga_Classes": [10, 15, 20, 25],
    "Pilates_Classes": [5, 10, 15, 20],
    "Number_of_Members": [20, 30, 40, 50]
}

df = pd.DataFrame(data)
```

# Obliczamy całkowitą liczbę członków

```python
total_members = df["Number_of_Members"].sum()
```

# Wyszukujemy wybraną kombinację zajęć

```python
selected_row = df[
    (df["Yoga_Classes"] == 15) &
    (df["Pilates_Classes"] == 10)
]
```

# Obliczamy prawdopodobieństwo

```python
probability = selected_row["Number_of_Members"].iloc[0] / total_members

print("Tabela danych:")
print(df)

print(f"\nŁączna liczba członków: {total_members}")
print(f"Prawdopodobieństwo: {probability:.4f}")
print(f"Prawdopodobieństwo procentowe: {probability * 100:.2f}%")
```
