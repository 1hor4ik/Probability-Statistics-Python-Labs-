```python
import pandas as pd
```

# Definiujemy dane podane w instrukcji

```python
data = {
    "Math_Score": [60, 70, 80, 90],
    "Science_Score": [70, 80, 85, 95],
    "Probability": [0.1, 0.15, 0.2, 0.25]
}

df = pd.DataFrame(data)
```

# Wyświetlamy rozkład wspólny

```python
print("Rozkład wspólny:")
print(df)
```

# Obliczamy rozkłady brzegowe przez sumowanie prawdopodobieństw

```python
marginal_math = df.groupby("Math_Score")["Probability"].sum()
marginal_science = df.groupby("Science_Score")["Probability"].sum()

print("\nRozkład brzegowy wyników z matematyki:")
print(marginal_math)

print("\nRozkład brzegowy wyników z nauk przyrodniczych:")
print(marginal_science)
```

# Sprawdzamy sumę prawdopodobieństw z instrukcji

```python
total_probability = df["Probability"].sum()

print(f"\nSuma prawdopodobieństw: {total_probability:.2f}")

if abs(total_probability - 1.0) < 1e-9:
    print("Dane tworzą znormalizowany rozkład prawdopodobieństwa.")
else:
    print("Uwaga: podane prawdopodobieństwa nie sumują się do 1.")
```
