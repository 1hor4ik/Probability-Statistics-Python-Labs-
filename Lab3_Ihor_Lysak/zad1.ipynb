```python
import pandas as pd
```

# Tworzymy ramkę danych na podstawie tabeli z instrukcji

```python
data = {
    "Action_Movie_Tickets": [8, 12, 18, 24],
    "Comedy_Movie_Tickets": [2, 5, 8, 15],
    "Number_of_Students": [10, 15, 20, 30]
}

df = pd.DataFrame(data)
```

# Obliczamy całkowitą liczbę studentów

```python
total_students = df["Number_of_Students"].sum()
```

# Wyszukujemy kombinację 18 biletów na filmy akcji i 8 na komedie

```python
selected_row = df[
    (df["Action_Movie_Tickets"] == 18) &
    (df["Comedy_Movie_Tickets"] == 8)
]
```

# Obliczamy prawdopodobieństwo wybranej kombinacji

```python
probability = selected_row["Number_of_Students"].iloc[0] / total_students

print("Tabela danych:")
print(df)

print(f"\nŁączna liczba studentów: {total_students}")
print(f"Prawdopodobieństwo: {probability:.4f}")
print(f"Prawdopodobieństwo procentowe: {probability * 100:.2f}%")
```
